# Linked Lists - Examples and Common LeetCode Problems

This notebook covers linked list fundamentals and common LeetCode problems involving linked lists.

## 1. Basic Linked List Node Definition

A linked list is a linear data structure where elements are stored in nodes, and each node points to the next node.

```python
class ListNode:
    """Definition for singly-linked list node."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

# Helper function to create a linked list from a list
def create_linked_list(values):
    """Create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head

# Helper function to print a linked list
def print_linked_list(head):
    """Print all values in the linked list."""
    values = []
    current = head
    while current:
        values.append(str(current.val))
        current = current.next
    print(" -> ".join(values) if values else "Empty list")

# Example usage
head = create_linked_list([1, 2, 3, 4, 5])
print_linked_list(head)
```

## 2. Reverse a Linked List (LeetCode 206)

Given the head of a singly linked list, reverse the list, and return the reversed list.

```python
def reverse_list(head):
    """Reverse a singly linked list iteratively."""
    prev = None
    current = head
    while current:
        next_temp = current.next  # Store next node
        current.next = prev       # Reverse the link
        prev = current            # Move prev forward
        current = next_temp       # Move current forward
    return prev

# Example
head = create_linked_list([1, 2, 3, 4, 5])
print("Original list:")
print_linked_list(head)

reversed_head = reverse_list(head)
print("Reversed list:")
print_linked_list(reversed_head)
```

## 3. Detect Cycle in a Linked List (LeetCode 141)

Given head, the head of a linked list, determine if the linked list has a cycle in it.

Using Floyd's Cycle Detection Algorithm (Tortoise and Hare).

```python
def has_cycle(head):
    """Detect if a linked list has a cycle using Floyd's algorithm."""
    if not head or not head.next:
        return False
    
    slow = head
    fast = head
    
    while fast and fast.next:
        slow = slow.next          # Move slow by 1
        fast = fast.next.next     # Move fast by 2
        if slow == fast:
            return True
    
    return False

# Example without cycle
head = create_linked_list([1, 2, 3, 4, 5])
print(f"Has cycle: {has_cycle(head)}")

# Example with cycle
head_with_cycle = create_linked_list([1, 2, 3, 4, 5])
# Create a cycle: 5 -> 3
current = head_with_cycle
third_node = None
count = 0
while current.next:
    count += 1
    if count == 3:
        third_node = current
    current = current.next
current.next = third_node  # Create cycle
print(f"Has cycle (after creating cycle): {has_cycle(head_with_cycle)}")
```

## 4. Find Middle of Linked List (LeetCode 876)

Given the head of a singly linked list, return the middle node of the linked list.

```python
def find_middle(head):
    """Find the middle node using slow and fast pointers."""
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow

# Example with odd number of nodes
head = create_linked_list([1, 2, 3, 4, 5])
print("List: 1 -> 2 -> 3 -> 4 -> 5")
middle = find_middle(head)
print(f"Middle node value: {middle.val}")

# Example with even number of nodes
head = create_linked_list([1, 2, 3, 4, 5, 6])
print("\nList: 1 -> 2 -> 3 -> 4 -> 5 -> 6")
middle = find_middle(head)
print(f"Middle node value (second middle): {middle.val}")
```

## 5. Merge Two Sorted Lists (LeetCode 21)

Merge two sorted linked lists and return it as a sorted list.

```python
def merge_two_lists(list1, list2):
    """Merge two sorted linked lists."""
    dummy = ListNode(0)
    current = dummy
    
    while list1 and list2:
        if list1.val <= list2.val:
            current.next = list1
            list1 = list1.next
        else:
            current.next = list2
            list2 = list2.next
        current = current.next
    
    # Attach remaining nodes
    current.next = list1 if list1 else list2
    
    return dummy.next

# Example
list1 = create_linked_list([1, 2, 4])
list2 = create_linked_list([1, 3, 4])
print("List 1: 1 -> 2 -> 4")
print("List 2: 1 -> 3 -> 4")

merged = merge_two_lists(list1, list2)
print("Merged list:")
print_linked_list(merged)
```

## 6. Remove Nth Node From End (LeetCode 19)

Given the head of a linked list, remove the nth node from the end of the list and return its head.

```python
def remove_nth_from_end(head, n):
    """Remove the nth node from the end using two pointers."""
    dummy = ListNode(0, head)
    first = dummy
    second = dummy
    
    # Move first pointer n+1 steps ahead
    for _ in range(n + 1):
        first = first.next
    
    # Move both pointers until first reaches the end
    while first:
        first = first.next
        second = second.next
    
    # Remove the nth node from end
    second.next = second.next.next
    
    return dummy.next

# Example
head = create_linked_list([1, 2, 3, 4, 5])
print("Original list:")
print_linked_list(head)

result = remove_nth_from_end(head, 2)
print("After removing 2nd node from end:")
print_linked_list(result)
```

## 7. Palindrome Linked List (LeetCode 234)

Given the head of a singly linked list, return true if it is a palindrome.

```python
def is_palindrome(head):
    """Check if a linked list is a palindrome."""
    if not head or not head.next:
        return True
    
    # Find the middle
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    
    # Reverse the second half
    prev = None
    while slow:
        next_temp = slow.next
        slow.next = prev
        prev = slow
        slow = next_temp
    
    # Compare both halves
    left, right = head, prev
    while right:
        if left.val != right.val:
            return False
        left = left.next
        right = right.next
    
    return True

# Example - palindrome
head = create_linked_list([1, 2, 2, 1])
print("List: 1 -> 2 -> 2 -> 1")
print(f"Is palindrome: {is_palindrome(head)}")

# Example - not a palindrome
head = create_linked_list([1, 2, 3])
print("\nList: 1 -> 2 -> 3")
print(f"Is palindrome: {is_palindrome(head)}")
```

## 8. Intersection of Two Linked Lists (LeetCode 160)

Given the heads of two singly linked-lists, return the node at which the two lists intersect.

```python
def get_intersection_node(headA, headB):
    """Find the intersection node of two linked lists."""
    if not headA or not headB:
        return None
    
    ptrA = headA
    ptrB = headB
    
    # Traverse both lists; when one ends, continue from the other's head
    while ptrA != ptrB:
        ptrA = ptrA.next if ptrA else headB
        ptrB = ptrB.next if ptrB else headA
    
    return ptrA

# Example
# Create two lists that intersect
# List A: 1 -> 2 \
#                  -> 3 -> 4 -> 5
# List B: 6 ------/
common = create_linked_list([3, 4, 5])
headA = ListNode(1, ListNode(2, common))
headB = ListNode(6, common)

intersection = get_intersection_node(headA, headB)
if intersection:
    print(f"Intersection at node with value: {intersection.val}")
else:
    print("No intersection")
```

## 9. Add Two Numbers (LeetCode 2)

You are given two non-empty linked lists representing two non-negative integers. The digits are stored in reverse order. Add the two numbers and return the sum as a linked list.

```python
def add_two_numbers(l1, l2):
    """Add two numbers represented as linked lists."""
    dummy = ListNode(0)
    current = dummy
    carry = 0
    
    while l1 or l2 or carry:
        val1 = l1.val if l1 else 0
        val2 = l2.val if l2 else 0
        
        total = val1 + val2 + carry
        carry = total // 10
        current.next = ListNode(total % 10)
        current = current.next
        
        l1 = l1.next if l1 else None
        l2 = l2.next if l2 else None
    
    return dummy.next

# Example: 342 + 465 = 807
# Represented as: [2,4,3] + [5,6,4] = [7,0,8]
l1 = create_linked_list([2, 4, 3])
l2 = create_linked_list([5, 6, 4])
print("Number 1 (342): 2 -> 4 -> 3")
print("Number 2 (465): 5 -> 6 -> 4")

result = add_two_numbers(l1, l2)
print("Sum (807):")
print_linked_list(result)
```

## 10. Delete Node in a Linked List (LeetCode 237)

Write a function to delete a node in a singly-linked list. You will not be given access to the head of the list, instead you will be given access to the node to be deleted directly.

```python
def delete_node(node):
    """Delete a node (except the tail) from a linked list, given only access to that node."""
    # Copy the value from the next node
    node.val = node.next.val
    # Skip the next node
    node.next = node.next.next

# Example
head = create_linked_list([4, 5, 1, 9])
print("Original list:")
print_linked_list(head)

# Get reference to node with value 5 (second node)
node_to_delete = head.next
print(f"Deleting node with value: {node_to_delete.val}")

delete_node(node_to_delete)
print("After deletion:")
print_linked_list(head)
```

## Summary

Key techniques for linked list problems:

1. **Two Pointers (Slow/Fast)**: Used for finding middle, detecting cycles, finding nth from end
2. **Dummy Node**: Simplifies edge cases when modifying the head
3. **Reversing**: Essential pattern for many problems
4. **Merging**: Combining sorted lists
5. **Carry Handling**: For arithmetic operations on linked lists
